# 📈 Exploratory Data Analysis (EDA)

> *Discovering hidden patterns, market concentration, and supply chain vulnerabilities through advanced exploratory analytics.*

---

## 🌍 Business Context

Critical mineral supply chains are highly sensitive to **production concentration**. When a small number of countries dominate global production, the likelihood of supply disruptions increases due to geopolitical tensions, export restrictions, natural disasters, or operational constraints.

To quantify this concentration, we use the **Herfindahl–Hirschman Index (HHI)**—a globally recognized market concentration metric widely applied in economics and antitrust analysis.

---

## 📚 What is the Herfindahl–Hirschman Index (HHI)?

The **HHI** measures how concentrated production is across participating countries.

<div align="center">

| HHI Score | Market Structure | Supply Risk |
|:---------:|:----------------|:-----------:|
| 🟢 **< 0.15** | Competitive Market | Low |
| 🟡 **0.15 – 0.25** | Moderately Concentrated | Medium |
| 🔴 **> 0.25** | Highly Concentrated | High |

</div>

> **Higher HHI values indicate greater dependence on a limited number of producers, increasing the vulnerability of global supply chains.**

---

## 🎯 Analysis Objectives

- 🌍 Examine the global distribution of critical mineral production.
- 📊 Measure market concentration using the **HHI**.
- ⚠️ Compare concentration levels with the computed **Supply Risk Score**.
- 🔍 Identify minerals and countries with elevated disruption potential.
- 📈 Generate business intelligence to support strategic procurement and investment decisions.

---

## 💡 Key Business Insight

> **If a mineral exhibits both a high HHI and a high Supply Risk Score, it represents a strategic vulnerability that may require supplier diversification, stockpiling, or long-term investment planning.**

---

# 🧠 3. Machine Learning Pipeline

> *Leveraging advanced machine learning to forecast future critical mineral supply chain disruptions through predictive analytics and explainable AI.*

---

# 🎯 Problem Statement

**Can we accurately predict whether a critical mineral supply chain will experience a disruption in the following year using current production, market concentration, trade, and geopolitical indicators?**

This predictive capability enables governments, mining companies, investors, and procurement teams to proactively identify vulnerabilities and implement risk mitigation strategies before disruptions occur.

---

# ⚙️ Methodology

## 🏗️ 1. Feature Engineering

To enhance predictive performance, advanced features were engineered from historical production and market data, including:

- 🔗 Interaction features *(e.g., `concentration × production growth`)*
- 📈 Growth rates and year-over-year changes
- 🌍 Market concentration indicators (HHI)
- 📦 Supply-demand imbalance metrics
- 🏛️ Geopolitical and governance factors
- ⏳ Lag and rolling statistical features

These engineered variables capture complex, non-linear relationships that influence future supply disruptions.

---

## 📅 2. Time-Series Data Split

To preserve temporal integrity and eliminate **data leakage**, the dataset was split chronologically:

<div align="center">

| Dataset | Years | Purpose |
|:--------:|:-----:|:--------|
| 📚 Training | **2015 – 2023** | Model Learning |
| 🧪 Testing | **2024 – 2025** | Out-of-Sample Evaluation |
| 🔮 Forecast | **2026** | Future Risk Prediction |

</div>

> **Chronological splitting ensures the model is evaluated on future observations, closely reflecting real-world forecasting scenarios.**

---

## 🤖 3. Machine Learning Model

The predictive engine utilizes an **XGBoost Classifier**, selected for its exceptional performance on structured tabular datasets.

### Why XGBoost?

- ⚡ High predictive accuracy
- 🌲 Captures complex non-linear relationships
- ⚖️ Robust handling of class imbalance
- 🧩 Built-in feature importance analysis
- 🚀 Fast training and scalable performance
- 📊 Strong generalization on real-world data

---

## 📊 4. Model Evaluation

Model performance is assessed using multiple complementary metrics:

| 📈 Metric | Purpose |
|:----------|:--------|
| 🎯 ROC–AUC | Measures overall discrimination ability |
| 📋 Classification Report | Precision, Recall & F1-Score |
| 🔲 Confusion Matrix | True vs. predicted classifications |
| ✅ Accuracy | Overall prediction correctness |

These evaluation metrics provide a comprehensive assessment of model reliability, robustness, and forecasting capability.

---

# 🚀 Expected Outcome

> **Develop a robust AI-driven forecasting model capable of identifying high-risk critical mineral supply chains before disruptions occur, enabling proactive procurement planning, strategic stockpiling, supplier diversification, and data-driven investment decisions.**

````python
# Feature Engineering
def engineer_features(dataframe):
    df = dataframe.copy()
    # Price volatility proxy within a mineral group
    df['price_volatility_proxy'] = df.groupby('mineral')['price_usd_per_tonne'].pct_change().fillna(0)
    # Interaction: High concentration + High demand growth = Extreme risk
    df['concentration_x_growth'] = df['hhi'] * (df['demand_growth_pct'] / 100)
    # Reserves pressure: Demand growth relative to years of reserves left
    df['reserves_pressure'] = df['demand_growth_pct'] / (df['years_of_reserves'] + 1)
    return df

df_train_test = engineer_features(df_modeling)

# Define Features and Target
target = 'disruption_next_year'
# 'disruption' is concurrent, we drop it to prevent target leakage
features = df_train_test.drop(columns=[target, 'disruption']).columns

# Time-based Split
train_df = df_train_test[df_train_test['year'] <= 2023]
test_df = df_train_test[df_train_test['year'] >= 2024]

X_train, y_train = train_df[features], train_df[target]
X_test, y_test = test_df[features], test_df[target]

# Preprocessing Pipeline
numeric_features = X_train.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = X_train.select_dtypes(exclude=[np.number]).columns.tolist()

preprocessor = ColumnTransformer([
    ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), numeric_features),
    ('cat', Pipeline([('imputer', SimpleImputer(strategy='constant', fill_value='missing')), 
                      ('onehot', OneHotEncoder(handle_unknown='ignore'))]), categorical_features)
])

# XGBoost Model (handling class imbalance via scale_pos_weight)
scale_pos_weight = (len(y_train) - sum(y_train)) / sum(y_train)
model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', xgb.XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=5, 
                                     scale_pos_weight=scale_pos_weight, random_state=42, eval_metric='logloss'))
])

model.fit(X_train, y_train)
print('✅ Model Training Complete')
````

````python
# Evaluation
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print("<--- Classification Report --->")
print(classification_report(y_test, y_pred))

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_proba)
auc_score = roc_auc_score(y_test, y_proba)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {auc_score:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.title('Receiver Operating Characteristic (ROC) - XGBoost', fontsize=14)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.show()
````
